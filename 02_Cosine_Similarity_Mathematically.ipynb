## 1. The Formula

$$
\text{cosine\_similarity}(A, B) = \frac{A \cdot B}{\|A\| \times \|B\|}
$$

- $A \cdot B = a_1 b_1 + a_2 b_2 + \dots + a_n b_n$ (dot product)
- $\|A\| = \sqrt{a_1^2 + a_2^2 + \dots + a_n^2}$ (magnitude)

| Result | Meaning |
|---|---|
| `1` | same direction (very similar) |
| `0` | perpendicular (unrelated) |
| `-1` | opposite direction |

```python
vector_A = [3, 1, 2]
vector_B = [1, 4, 2]

print("Vector A:", vector_A)
print("Vector B:", vector_B)
```

## 2. Step 1: Dot Product

```python
def dot_product(A, B):
    total = 0
    for a, b in zip(A, B):
        total += a * b
    return total

dp = dot_product(vector_A, vector_B)
print("Dot product:", dp, "  (3*1 + 1*4 + 2*2 =", 3*1 + 1*4 + 2*2, ")")
```

## 3. Step 2: Magnitude

```python
import math

def magnitude(A):
    return math.sqrt(sum(a ** 2 for a in A))

mag_A = magnitude(vector_A)
mag_B = magnitude(vector_B)

print("|A| =", round(mag_A, 4))
print("|B| =", round(mag_B, 4))
```

## 4. Step 3: Cosine Similarity from Scratch

```python
def cosine_similarity_manual(A, B):
    return dot_product(A, B) / (magnitude(A) * magnitude(B))

result_manual = cosine_similarity_manual(vector_A, vector_B)
print("Cosine similarity (manual):", round(result_manual, 4))
```

## 5. Verify with NumPy

```python
import numpy as np

A = np.array(vector_A)
B = np.array(vector_B)

result_numpy = np.dot(A, B) / (np.linalg.norm(A) * np.linalg.norm(B))
print("Cosine similarity (NumPy):", round(result_numpy, 4))
```

## 6. Verify with scikit-learn

```python
from sklearn.metrics.pairwise import cosine_similarity

# sklearn expects 2D arrays of shape (n_samples, n_features)
result_sklearn = cosine_similarity(A.reshape(1, -1), B.reshape(1, -1))[0][0]
print("Cosine similarity (scikit-learn):", round(result_sklearn, 4))
```

```python
print("Manual       :", round(result_manual, 6))
print("NumPy        :", round(float(result_numpy), 6))
print("scikit-learn :", round(float(result_sklearn), 6))

assert np.isclose(result_manual, result_numpy) and np.isclose(result_numpy, result_sklearn)
print("\nAll three methods match!")
```

## 7. Edge Cases to Build Intuition

```python
examples = {
    "Identical vectors":              ([2, 3, 4], [2, 3, 4]),
    "Opposite vectors":               ([2, 3, 4], [-2, -3, -4]),
    "Perpendicular vectors":          ([1, 0], [0, 1]),
    "Same direction, different size": ([1, 2, 3], [10, 20, 30]),
    "Partly similar":                 ([1, 2, 3], [3, 2, 1]),
}

for name, (a, b) in examples.items():
    print(f"{name:32s} -> {cosine_similarity_manual(a, b):.4f}")
```

Scaling a vector does **not** change its cosine similarity, because only direction matters. This is why it suits text embeddings.

## 8. Cosine Similarity vs Euclidean Distance
Euclidean distance is affected by vector length, cosine similarity is not.

```python
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])   # same direction as a, much longer

print("Euclidean distance :", round(np.linalg.norm(a - b), 3), "(looks far apart)")
print("Cosine similarity  :", round(cosine_similarity_manual(a, b), 3), "(same direction)")
```

## 9. Cosine Distance (used by vector databases)

```python
def cosine_distance(A, B):
    return 1 - cosine_similarity_manual(A, B)

print("Distance between A and B       :", round(cosine_distance(vector_A, vector_B), 4))
print("Distance between A and itself  :", round(cosine_distance(vector_A, vector_A), 4))
```

## 11. Practice Exercises (Solved)

```python
# Exercise 1: cosine similarity using list comprehensions instead of loops
def cosine_similarity_compact(A, B):
    dot = sum(a * b for a, b in zip(A, B))
    return dot / (math.sqrt(sum(a * a for a in A)) * math.sqrt(sum(b * b for b in B)))

print("Exercise 1:", round(cosine_similarity_compact(vector_A, vector_B), 4))

# Exercise 2: [1,1,1] vs [5,5,5]. Prediction: same direction, so exactly 1.0
print("Exercise 2:", round(cosine_similarity_manual([1, 1, 1], [5, 5, 5]), 4))
```

## Key Takeaways

| Term | Meaning |
|---|---|
| Dot product | Sum of element-wise products |
| Magnitude | Length of a vector |
| Cosine similarity | Dot product divided by the product of magnitudes |
| Range | -1 (opposite) to 1 (same direction) |
| Cosine distance | `1 - cosine similarity`, what vector databases usually report |
| Why it matters | Compares direction (meaning) and ignores length |
